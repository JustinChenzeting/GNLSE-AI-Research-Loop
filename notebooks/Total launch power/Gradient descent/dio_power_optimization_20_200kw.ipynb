# Optimize LP01 input–output distortion over 20–200 kW

Vary input peak power over **20–200 kW**, starting at **100 kW**, to minimize the change
between the input pulse and nonlinear output. Keep the original **200 fs
Gaussian**, **LP01 launch and score**, **three propagated modes**, **2.86 m
step-index fiber**, and **1379.2 nm carrier**.

\[
F_{\mathrm{IO},01}=\frac{\left|\int A_{\mathrm{in},1}^{*}(0,t)
A_{\mathrm{NL},1}(L,t)\,dt\right|^2}
{\left(\int |A_{\mathrm{in},1}(0,t)|^2dt\right)
 \left(\int |A_{\mathrm{NL},1}(L,t)|^2dt\right)},
\qquad D_{\mathrm{IO},01}=1-F_{\mathrm{IO},01}.
\]

**Lower DIO is better.** Evaluate both fields on the same time samples with
**no alignment, time shift, spectral filter, or phase correction**. The complex
overlap is insensitive to a single constant complex scale factor. All three
modes propagate and may couple; this notebook retains the original LP01-only
score. Thus the modal sum in the general DIO definition is restricted to LP01.
The matched linear output appears only as a diagnostic in the final figures.

For every candidate, require **at least 50% LP01 photon retention**. Divide its unfiltered LP01 output photon count by its LP01 input photon count at the same launch power.

Report the gradient-descent endpoint, the lowest sampled D_IO, and the final
feasible selection separately. Final selection compares the gradient endpoint
with the lowest feasible sweep candidate on the 1 mm grid.

Run in order with **GNLSE (Julia)**. This is an independent
DIO run; its cache and saved results are separate from the source notebook.

## 1. Choose the control and compute budget

Use \(u=P/(100\,\mathrm{kW})\), **20–200 kW** bounds, and at least **50%**
of each launch’s own LP01 input photons. Search with 10 mm propagation steps; perform
one finer 1 mm propagation check for the baseline and selected result.

The finite-difference spacing remains ±200 W (clipped at the power bounds).
Each new objective evaluation runs **one nonlinear forward solve**. Keep the
original learning rate, 20-iteration limit, and 80-new-evaluation optimizer
budget. Rerun from the top after changing settings so the caches are rebuilt.

```julia
reference_power_W = 100_000.0
power_bounds_W = (20_000.0, 200_000.0)
sweep_step_W = 2_000.0           # 91 points across 20–200 kW
minimum_brightness_fraction = 0.50
fd_step = 0.002                  # ±200 W when differentiating with respect to u
learning_rate = 0.10
max_iterations = 20
max_optimizer_evaluations = 80
gradient_tolerance = 1e-4        # DIO per unit u
max_power_change_W = 10_000.0
search_dz = 0.01                # m
validation_dz = 0.001           # m

dio_step_tolerance = 2e-4  # absolute DIO difference, not dB
launch_coefficients = [1.0, 0.0, 0.0]

settings = Dict(
    "control" => "input Gaussian peak power (W)",
    "objective" => "minimize LP01 DIO = 1 - input/nonlinear-output complex field fidelity",
    "reference_power_W" => reference_power_W,
    "power_bounds_W" => collect(power_bounds_W),
    "sweep_step_W" => sweep_step_W,
    "minimum_brightness_fraction" => minimum_brightness_fraction,
    "finite_difference_step_u" => fd_step,
    "learning_rate" => learning_rate,
    "max_iterations" => max_iterations,
    "max_optimizer_evaluations" => max_optimizer_evaluations,
    "gradient_tolerance" => gradient_tolerance,
    "max_power_change_W" => max_power_change_W,
    "search_dz_m" => search_dz,
    "validation_dz_m" => validation_dz,
    "Nt" => 2048, "time_window_ps" => 10.0, "pulse_fwhm_ps" => 0.2,
    "fiber_length_m" => 2.86, "carrier_m" => 1379.2e-9,
    "core_radius_m" => 52.5e-6, "NA" => 0.1, "num_modes" => 3,
    "objective_mode" => 1, "brightness_modes" => [1], "objective_filter" => "none",
    "objective_reference" => "original candidate input on unchanged time samples; no alignment or shift",
    "linear_diagnostic" => "same candidate input and fiber; n2 = 0; final plots only",
    "brightness_definition" => "unfiltered LP01 output photons divided by LP01 input photons at the same power",
    "dio_step_tolerance" => dio_step_tolerance,
    "launch_coefficients" => launch_coefficients,
    "cp_target_error" => 1e-3, "cp_seed" => Int(0x4d4d474e),
    "julia_version" => string(VERSION),
    "manifest_sha256" => bytes2hex(sha256(read(joinpath(ROOT, "Manifest.toml")))),
)
open(joinpath(run_dir, "settings.toml"), "w") do io
    TOML.print(io, settings)
end
```

## 2. Build the fixed fiber

Use the original three-mode fiber and CP compression. The nonlinear parameter
set supplies the objective; the n2 = 0 parameter set is used only for final
baseline/selected diagnostic plots. No dense-versus-CP or extra time-grid sweep
is run here.

Photon retention counts LP01 photons over the full spectrum at the input and output. All-mode photon count
and modal energy fractions remain physical diagnostics.

```julia
time_grid = TimeGrid(settings["Nt"], settings["time_window_ps"])
domain = MMGNLSEDomain(degrees_of_freedom(:time, :space), time_grid)
properties = compute_fiber_properties(
    StepIndex(settings["carrier_m"], settings["core_radius_m"], settings["NA"]),
    Silica(), settings["num_modes"];
    polarization=:scalar, mode_basis=:LP, beta_order=5)
S_cp = cp_compress(properties.S; error=settings["cp_target_error"], seed=settings["cp_seed"])
parameters_cp = MMGNLSEParameters(domain;
    length=settings["fiber_length_m"], beta=properties.beta, S=S_cp,
    n2=properties.n2, omega0=properties.omega0, raman=properties.raman)


parameters_linear = MMGNLSEParameters(domain;
    length=settings["fiber_length_m"], beta=properties.beta, S=S_cp,
    n2=0.0, omega0=properties.omega0, raman=properties.raman)
lp01_photons = SpectralPhotonNumber(filter=ones(settings["Nt"]), modes=[1])
all_photons = SpectralPhotonNumber(filter=ones(settings["Nt"]), modes=:all)
@printf("Three propagated modes; LP01 launch and score. CP error: %.3g\n", S_cp.relative_error)
println("Mode labels: ", properties.modes.labels)
```

## 3. Turn the DIO calculation into a function

The common time-spacing factors cancel in the normalized overlap. Compare the
**original input and nonlinear output**, including their complex phase, at
identical time indices. Keep the original LP01-only score and reject zero or
nonfinite field norms. Algebra checks establish self-overlap, invariance to a
constant complex factor, and zero overlap for orthogonal samples.

Cache each nonlinear evaluation and its endpoint fields in this independent
run directory. Linear propagation is deferred until the final plots.

```julia
evaluation_cache = Dict{Tuple{Float64,Float64},NamedTuple}()
field_cache = Dict{Tuple{Float64,Float64},NamedTuple}()
evaluations = NamedTuple[]

let a = zeros(ComplexF64, 4, 3, 1)
    a[1, 1, 1] = 1
    @assert isapprox(lp01_fidelity(a, a), 1.0; atol=1e-12)
    @assert isapprox(lp01_fidelity(a, 2exp(0.7im).*a), 1.0; atol=1e-12)
    b = copy(a)
    b[:, 2, 1] .= 3  # Changes in other modes do not enter this objective.
    @assert isapprox(lp01_fidelity(a, b), 1.0; atol=1e-12)
    b = zeros(ComplexF64, size(a))
    b[2, 1, 1] = 1
    @assert lp01_fidelity(a, b) == 0.0
    rejected_zero = try
        lp01_fidelity(a, zero(a))
        false
    catch err
        err isa ArgumentError
    end
    @assert rejected_zero
end
println("LP01 overlap definition checks passed.")
```

gradient step the best point across all evaluations.

The sweep has 2 kW spacing and does not certify the continuous global minimum.

```julia
scan = [dio_at(P) for P in power_bounds_W[1]:sweep_step_W:power_bounds_W[2]]
save_rows("power_sweep.csv", scan)
feasible_scan = filter(r -> r.photon_retention >= minimum_brightness_fraction, scan)
@assert !isempty(feasible_scan)
sweep_minimum = scan[argmin([r.D_IO for r in scan])]
sweep_feasible_minimum = feasible_scan[argmin([r.D_IO for r in feasible_scan])]

# Reuse already checked powers; only new candidate powers need a fine solve.
sweep_feasible_fine = same_power(sweep_feasible_minimum,gd_fine) ? gd_fine :
    dio_at(sweep_feasible_minimum.power_W;dz=validation_dz)
sweep_minimum_fine = same_power(sweep_minimum,gd_fine) ? gd_fine :
    same_power(sweep_minimum,sweep_feasible_fine) ? sweep_feasible_fine :
    dio_at(sweep_minimum.power_W;dz=validation_dz)
feasible_sweep_step_error = abs(sweep_feasible_minimum.D_IO-sweep_feasible_fine.D_IO)
sweep_minimum_step_error = abs(sweep_minimum.D_IO-sweep_minimum_fine.D_IO)
@assert feasible_sweep_step_error < dio_step_tolerance
@assert sweep_minimum_step_error < dio_step_tolerance
sweep_candidate_passes_brightness = sweep_feasible_fine.photon_retention >= minimum_brightness_fraction

selected_search = gd_search
best_fine = gd_fine
selected_source = "gradient-descent endpoint"
if sweep_candidate_passes_brightness && sweep_feasible_fine.D_IO < gd_fine.D_IO-1e-12
    selected_search = sweep_feasible_minimum
    best_fine = sweep_feasible_fine
    selected_source = "lowest feasible sweep sample"
elseif same_power(sweep_feasible_minimum,gd_fine)
    selected_source = "gradient endpoint and lowest feasible sweep sample coincide"
end
best_power_W = best_fine.power_W
best_step_error = abs(selected_search.D_IO-best_fine.D_IO)
@assert best_fine.photon_retention >= minimum_brightness_fraction && best_step_error < dio_step_tolerance
improvement_DIO = baseline_fine.D_IO-best_fine.D_IO
merge!(summary,Dict(
    "selected_source"=>selected_source,
    "best_power_W"=>best_power_W,"best_D_IO"=>best_fine.D_IO,"best_F_IO"=>best_fine.F_IO,
    "best_input_photons"=>best_fine.input_photons,"best_photons"=>best_fine.photons,"brightness_ratio"=>best_fine.photon_retention,
    "improvement_DIO"=>improvement_DIO,"relative_DIO_reduction"=>improvement_DIO/baseline_fine.D_IO,
    "best_step_error"=>best_step_error,
    "best_lp01_energy_fraction"=>best_fine.lp01_energy_fraction,
    "best_mode2_energy_fraction"=>best_fine.mode2_energy_fraction,
    "best_mode3_energy_fraction"=>best_fine.mode3_energy_fraction,
    "best_output_input_energy_ratio"=>best_fine.output_input_energy_ratio,
    "sweep_points"=>length(scan),"feasible_sweep_points"=>length(feasible_scan),
    "sweep_minimum_power_W"=>sweep_minimum.power_W,
    "sweep_minimum_D_IO_search"=>sweep_minimum.D_IO,
    "sweep_minimum_D_IO_fine"=>sweep_minimum_fine.D_IO,
    "sweep_minimum_F_IO_fine"=>sweep_minimum_fine.F_IO,
    "sweep_minimum_brightness_ratio"=>sweep_minimum_fine.photon_retention,
    "sweep_minimum_passes_brightness"=>sweep_minimum_fine.photon_retention>=minimum_brightness_fraction,
    "sweep_minimum_step_error"=>sweep_minimum_step_error,
    "sweep_feasible_minimum_power_W"=>sweep_feasible_minimum.power_W,
    "sweep_feasible_minimum_D_IO_fine"=>sweep_feasible_fine.D_IO,
    "sweep_feasible_minimum_passes_fine_brightness"=>sweep_candidate_passes_brightness,
    "sweep_feasible_minimum_step_error"=>feasible_sweep_step_error,
    "total_objective_evaluations"=>length(evaluations),
    "global_minimum_established"=>false,
    "validation_scope"=>"LP01 overlap/brightness; fine baseline, gradient endpoint, lowest feasible sweep candidate and lowest sampled D_IO; coincident powers reused"))
open(joinpath(run_dir,"summary.toml"),"w") do io
    TOML.print(io,summary)
end
save_rows("selection_comparison.csv",[
    (;label="baseline",power_W=baseline_fine.power_W,D_IO=baseline_fine.D_IO,F_IO=baseline_fine.F_IO,
      input_photons=baseline_fine.input_photons,output_photons=baseline_fine.photons,
      photon_retention=baseline_fine.photon_retention,brightness_ratio=baseline_fine.photon_retention,feasible=baseline_fine.photon_retention>=minimum_brightness_fraction,dz=baseline_fine.dz),
    (;label="gradient_endpoint",power_W=gd_fine.power_W,D_IO=gd_fine.D_IO,F_IO=gd_fine.F_IO,
      input_photons=gd_fine.input_photons,output_photons=gd_fine.photons,
      photon_retention=gd_fine.photon_retention,brightness_ratio=gd_fine.photon_retention,feasible=gd_fine.photon_retention>=minimum_brightness_fraction,dz=gd_fine.dz),
    (;label="lowest_sweep_D_IO",power_W=sweep_minimum_fine.power_W,D_IO=sweep_minimum_fine.D_IO,F_IO=sweep_minimum_fine.F_IO,
      input_photons=sweep_minimum_fine.input_photons,output_photons=sweep_minimum_fine.photons,
      photon_retention=sweep_minimum_fine.photon_retention,brightness_ratio=sweep_minimum_fine.photon_retention,feasible=sweep_minimum_fine.photon_retention>=minimum_brightness_fraction,dz=sweep_minimum_fine.dz),
    (;label="selected_feasible",power_W=best_fine.power_W,D_IO=best_fine.D_IO,F_IO=best_fine.F_IO,
      input_photons=best_fine.input_photons,output_photons=best_fine.photons,
      photon_retention=best_fine.photon_retention,brightness_ratio=best_fine.photon_retention,feasible=true,dz=best_fine.dz)])

fig, axes = subplots(1,2;figsize=(14,4.7),constrained_layout=true)
ax=axes[1]
ax.plot([r.power_W/1000 for r in scan],[r.D_IO for r in scan],"o-";
    color="#366b9e",markersize=3,label="Full sweep: 2 kW spacing")
bad=filter(r->r.photon_retention<minimum_brightness_fraction,scan)
if !isempty(bad)
    ax.scatter([r.power_W/1000 for r in bad],[r.D_IO for r in bad];
        color="gray",marker="x",s=35,label="Below LP01 photon floor",zorder=4)
end
h=optimization.history
ax.plot([r.power_W/1000 for r in h],[r.D_IO for r in h],"s--";
    color="#cf7c32",markersize=4,label="Accepted gradient steps")
for i in 2:length(h)
    ax.annotate("";xy=(h[i].power_W/1000,h[i].D_IO),xytext=(h[i-1].power_W/1000,h[i-1].D_IO),
        arrowprops=Dict("arrowstyle"=>"-|>","color"=>"#cf7c32","lw"=>1))
end
ax.annotate("Start";xy=(h[1].power_W/1000,h[1].D_IO),xytext=(6,8),textcoords="offset points",fontsize=8)
ax.annotate("GD endpoint";xy=(gd_search.power_W/1000,gd_search.D_IO),xytext=(8,12),textcoords="offset points",fontsize=8)
ax.scatter([selected_search.power_W/1000],[selected_search.D_IO];
    color="#2b8c5e",marker="*",s=140,label="Selected feasible result",zorder=6)
if !same_power(sweep_minimum,selected_search)
    ax.scatter([sweep_minimum.power_W/1000],[sweep_minimum.D_IO];
        color="#8057a6",marker="D",s=55,label="Lowest sampled D_IO",zorder=5)
end
fine_rows=[baseline_fine,best_fine,sweep_minimum_fine]
ax.scatter([r.power_W/1000 for r in fine_rows],[r.D_IO for r in fine_rows];
    color="black",marker="x",s=45,label="1 mm checks",zorder=7)
ax.set_xlabel("Input peak power (kW)");ax.set_ylabel("LP01 D_IO")
ax.set_title("20–200 kW; LP01 launch and score")
ax.legend(;fontsize=7.5,loc="best")
axes[2].plot([r.iteration for r in h],[r.D_IO for r in h],"o-";color="#cf7c32",markersize=4)
axes[2].scatter([h[1].iteration,h[end].iteration],[baseline_fine.D_IO,gd_fine.D_IO];
    color="black",marker="x",s=45,label="1 mm checks")
axes[2].set_xlabel("Accepted gradient iteration");axes[2].set_ylabel("LP01 D_IO")
axes[2].set_title("Gradient descent stop:\n"*optimization.status;fontsize=10)
axes[2].legend(;fontsize=8)
for ax in axes
    ax.grid(;alpha=0.2);ax.spines["top"].set_visible(false);ax.spines["right"].set_visible(false)
end
savefig(joinpath(run_dir,"dio_power_optimization.png");dpi=180)
savefig(joinpath(run_dir,"dio_power_optimization.pdf"));display(fig);close(fig)

@printf("GD endpoint: %.6f kW, 1 mm D_IO %.8f; stop: %s\n",gd_fine.power_W/1000,gd_fine.D_IO,optimization.status)
@printf("Lowest sampled D_IO: %.3f kW, 1 mm D_IO %.8f; LP01 input photons retained %.5f%%; feasible=%s\n",
    sweep_minimum_fine.power_W/1000,sweep_minimum_fine.D_IO,
    100sweep_minimum_fine.photon_retention,string(summary["sweep_minimum_passes_brightness"]))
@printf("Selected feasible result: %.6f kW, 1 mm D_IO %.8f; LP01 input photons retained %.5f%%\n",
    best_power_W/1000,best_fine.D_IO,100best_fine.photon_retention)
println("Selection source: ",selected_source)
println("Final input/output plots show the selected feasible result.")
@printf("Total objective evaluations (baseline, search, sweep, fine checks): %d\n",length(evaluations))
```

## 9. Take a peek at the spatial modes, input and output

First, display the three fixed mode shapes in the same style as the original
notebook. Red and blue indicate opposite signs of the mode's field amplitude.
These shapes form the spatial basis used to propagate the pulse.
The dashed circle is the fiber core; both axes are positions in micrometres.

```julia
x = properties.modes.x
y = properties.modes.y
modes = properties.modes.fields  # axes: y, x, mode
mode_names = ("LP01", "LP11 — cos orientation", "LP11 — sin orientation")

let fig = figure(; figsize=(18, 6), constrained_layout=true)
    for ii = 1:3
        ax = subplot(1, 3, ii)
        # Normalize amplitude only for displaying the fixed mode shapes.
        profile = real.(modes[:, :, ii])
        profile = profile ./ maximum(abs, profile)
        mesh = ax.pcolormesh(x.*1e6, y.*1e6, profile;
            cmap="seismic", vmin=-1, vmax=1, shading="auto", rasterized=true)
        ax.set_title("Mode $ii: "*mode_names[ii])
        ax.set_xlabel("x (μm)")
        ax.set_ylabel("y (μm)")
        ax.set_aspect("equal")
        theta = range(0, 2pi; length=181)
        ax.plot(settings["core_radius_m"].*1e6.*cos.(theta),
            settings["core_radius_m"].*1e6.*sin.(theta);
            color="#555555", linestyle="--", linewidth=0.8)
        fig.colorbar(mesh; ax, shrink=0.75, label="Relative field amplitude")
    end
    fig.suptitle("The three spatial modes — dashed line marks the fiber core")
    savefig(joinpath(run_dir, "spatial_modes.png"); dpi=160)
    savefig(joinpath(run_dir, "spatial_modes.pdf"))
    display(fig)
    close(fig)
end
npzwrite(joinpath(run_dir, "spatial_mode_fields.npz"),
    Dict("x_m"=>x, "y_m"=>y, "modes_y_x_m"=>modes))
```
